```python
# 1. Rapikan nama kolom
df.columns = (
    df.columns
    .str.replace("ï»¿", "", regex=False)
    .str.strip()
)

# 2. Periksa nama kolom setelah diperbaiki
print("Nama kolom setelah diperbaiki:")
print(df.columns.tolist())

# 3. Hapus kolom yang seluruh isinya kosong
kolom_kosong = [
    kolom for kolom in df.columns
    if df[kolom].isna().all()
]

print("\nKolom yang dihapus:", kolom_kosong)

df = df.drop(columns=kolom_kosong)

# 4. Buat salinan teks asli
df["original_text"] = df["text"].copy()

print("\nUkuran dataset setelah dirapikan:", df.shape)

display(df.head())
```

```python
# Periksa tanggal postingan
df["createdAt"] = pd.to_datetime(
    df["createdAt"],
    errors="coerce",
    utc=True
)

# Ubah ke zona waktu Indonesia
df["tanggal"] = (
    df["createdAt"]
    .dt.tz_convert("Asia/Jakarta")
)

print("Tanggal paling awal:")
print(df["tanggal"].min())

print("\nTanggal paling akhir:")
print(df["tanggal"].max())

print("\nJumlah data berdasarkan bahasa:")
display(df["lang"].value_counts(dropna=False))

print("\nContoh teks postingan:")
display(df[["tanggal", "lang", "text"]].head(10))
```

```python
print("Jumlah ID duplikat:")
print(df["id"].duplicated().sum())

# Hapus jika ada ID yang benar-benar duplikat
df = df.drop_duplicates(
    subset=["id"],
    keep="first"
)

print("Jumlah data setelah pemeriksaan ID:", len(df))
```

```python
# 1. Mengubah createdAt menjadi format tanggal
df["createdAt"] = pd.to_datetime(
    df["createdAt"],
    errors="coerce",
    utc=True
)

# 2. Mengubah waktu ke zona WIB
df["tanggal"] = df["createdAt"].dt.tz_convert(
    "Asia/Jakarta"
)

# 3. Melihat rentang tanggal
print("Tanggal paling awal:")
print(df["tanggal"].min())

print("\nTanggal paling akhir:")
print(df["tanggal"].max())

# 4. Melihat distribusi bahasa
print("\nJumlah data berdasarkan bahasa:")
display(df["lang"].value_counts(dropna=False))

# 5. Melihat 10 contoh postingan
display(
    df[["tanggal", "lang", "text"]].head(10)
)
```

```python
# Menentukan periode penelitian
tanggal_awal = pd.Timestamp(
    "2026-09-13 00:00:00",
    tz="Asia/Jakarta"
)

tanggal_akhir = pd.Timestamp(
    "2026-10-14 23:59:59",
    tz="Asia/Jakarta"
)

# Filter berdasarkan tanggal
df = df[
    (df["tanggal"] >= tanggal_awal) &
    (df["tanggal"] <= tanggal_akhir)
].copy()

print("Jumlah data setelah filter tanggal:", len(df))

print("Tanggal paling awal:", df["tanggal"].min())
print("Tanggal paling akhir:", df["tanggal"].max())
```

```python
# Simpan salinan data sebelum proses cleaning
df_cleaning = df.copy()

# Pastikan kolom teks tidak kosong
df_cleaning["text"] = df_cleaning["text"].fillna("").astype(str)

print("Jumlah data sebelum cleaning:", len(df_cleaning))
```

```python
import re

pola_relevan = (
    r"10[.\-/ ]?10|"
    r"tanggal kembar|"
    r"promo|promosi|"
    r"diskon|discount|"
    r"voucher|kupon|"
    r"flash sale|"
    r"gratis ongkir|ongkir|"
    r"shopee|tokopedia|lazada|blibli|"
    r"marketplace|belanja online"
)

df_cleaning["relevansi_awal"] = df_cleaning["text"].str.contains(
    pola_relevan,
    case=False,
    regex=True,
    na=False
)

print("Postingan terindikasi relevan:")
print(df_cleaning["relevansi_awal"].value_counts())

display(
    df_cleaning[["text", "relevansi_awal"]].head(20)
)
```

```python
#Text Cleaning
#Buat Fungsi Pembersihan Awal
import re
import unicodedata

def clean_text(text):
    text = str(text)

    # 1. Normalisasi karakter dan huruf kecil
    text = unicodedata.normalize("NFKC", text)
    text = text.lower()

    # 2. Hapus URL dan email
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)
    text = re.sub(r"\b[\w.+-]+@[\w.-]+\.[a-zA-Z]{2,}\b", " ", text)

    # 3. Hapus mention akun, tetapi pertahankan hashtag sebagai kata
    text = re.sub(r"(?<!\w)@\w+", " ", text)
    text = text.replace("#", "")

    # 4. Pertahankan angka, titik, dan persen untuk informasi promo
    # Bersihkan simbol lain
    text = re.sub(r"[^\w\s.%/-]", " ", text)

    # 5. Rapikan spasi
    text = re.sub(r"\s+", " ", text).strip()

    return text

df_cleaning["clean_text"] = df_cleaning["text"].apply(clean_text)

display(
    df_cleaning[["text", "clean_text"]].head(15)
)
```

```python
normalisasi = {
    "gak": "tidak",
    "ga": "tidak",
    "nggak": "tidak",
    "enggak": "tidak",
    "gk": "tidak",
    "bgt": "banget",
    "bngt": "banget",
    "murce": "murah",
    "udh": "sudah",
    "udah": "sudah",
    "blm": "belum",
    "tdk": "tidak",
    "krn": "karena",
    "dgn": "dengan",
    "yg": "yang",
    "aja": "saja"
}

def normalize_slang(text):
    words = text.split()
    words = [normalisasi.get(word, word) for word in words]
    return " ".join(words)

df_cleaning["clean_text"] = df_cleaning["clean_text"].apply(normalize_slang)

display(
    df_cleaning[["text", "clean_text"]].head(15)
)
```

```python
df_cleaning.to_csv(
    "data_tanggal_kembar_cleaning_awal.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Dataset cleaning awal berhasil disimpan.")
```
