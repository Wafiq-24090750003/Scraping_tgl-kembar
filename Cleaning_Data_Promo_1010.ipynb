```python
#Text Cleaning
#Buat Fungsi Pembersihan Awal
import re
import unicodedata

def clean_text(text):
    text = str(text)

    # 1. Normalisasi karakter dan huruf kecil
    text = unicodedata.normalize("NFKC", text)
    text = text.lower()

    # 2. Hapus URL dan email
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)
    text = re.sub(r"\b[\w.+-]+@[\w.-]+\.[a-zA-Z]{2,}\b", " ", text)

    # 3. Hapus mention akun, tetapi pertahankan hashtag sebagai kata
    text = re.sub(r"(?<!\w)@\w+", " ", text)
    text = text.replace("#", "")

    # 4. Pertahankan angka, titik, dan persen untuk informasi promo
    # Bersihkan simbol lain
    text = re.sub(r"[^\w\s.%/-]", " ", text)

    # 5. Rapikan spasi
    text = re.sub(r"\s+", " ", text).strip()

    return text

df_cleaning["clean_text"] = df_cleaning["text"].apply(clean_text)

display(
    df_cleaning[["text", "clean_text"]].head(15)
)
```

```python

import re
import unicodedata

def clean_text(text):
    text = str(text)
    text = unicodedata.normalize("NFKC", text)
    text = text.lower()

    # Hapus URL dan email
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)
    text = re.sub(
        r"\b[\w.+-]+@[\w.-]+\.[a-zA-Z]{2,}\b",
        " ", text
    )

    # Hapus mention
    text = re.sub(r"(?<!\w)@\w+", " ", text)

    # Hilangkan karakter pengganti dan karakter kontrol
    text = re.sub(r"[\ufffd]", " ", text)
    text = "".join(
        ch for ch in text
        if unicodedata.category(ch) not in ("Cc", "Cs")
    )

    # Hilangkan simbol emoji, tetapi pertahankan huruf,
    # angka, spasi, serta tanda yang berguna untuk promosi
    text = re.sub(r"[^\w\s.%/-]", " ", text)

    # Rapikan spasi
    text = re.sub(r"\s+", " ", text).strip()
    return text

df_cleaning["clean_text"] = (
    df_cleaning["original_text"]
    .fillna("")
    .apply(clean_text)
)

print("Teks kosong:", (df_cleaning["clean_text"].str.strip() == "").sum())
display(df_cleaning[["original_text", "clean_text"]].head(15))
```

```python

import re
import unicodedata

def clean_text(text):
    text = str(text)
    text = unicodedata.normalize("NFKC", text).lower()

    # Ganti baris baru dan tab dengan spasi
    text = re.sub(r"[\r\n\t]+", " ", text)

    # Hapus URL sebelum membersihkan tanda baca
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)

    # Hapus email dan mention
    text = re.sub(
        r"\b[\w.+-]+@[\w.-]+\.[a-zA-Z]{2,}\b", " ", text
    )
    text = re.sub(r"(?<!\w)@\w+", " ", text)

    # Hapus karakter kontrol dan pengganti yang rusak
    text = "".join(
        ch for ch in text
        if unicodedata.category(ch) not in ("Cc", "Cs")
    )
    text = text.replace("\ufffd", " ")

    # Hapus simbol, tetapi pertahankan angka, persen,
    # titik, garis miring, dan tanda minus yang berguna
    text = re.sub(r"[^\w\s.%/-]", " ", text)

    # Rapikan spasi
    text = re.sub(r"\s+", " ", text).strip()
    return text

df_cleaning["clean_text"] = (
    df_cleaning["original_text"]
    .fillna("")
    .apply(clean_text)
)

display(df_cleaning[["original_text", "clean_text"]].head(15))
```

```python
import re
import unicodedata
import pandas as pd

def clean_text_v2(text):
    if pd.isna(text):
        return ""

    text = unicodedata.normalize("NFKC", str(text)).lower()

    # Hapus URL dan email
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)
    text = re.sub(
        r"\b[\w.+-]+@[\w.-]+\.[a-zA-Z]{2,}\b",
        " ",
        text
    )

    # Hapus mention, tetapi pertahankan hashtag sebagai kata
    text = re.sub(r"(?<!\w)@\w+", " ", text)
    text = re.sub(r"#(\w+)", r"\1", text)

    # Hapus karakter kontrol dan simbol emoji yang tersisa.
    # Huruf, angka, tanda baca, dan spasi tetap dipertahankan.
    hasil = []
    for ch in text:
        kategori = unicodedata.category(ch)
        if kategori[0] in ("L", "N", "P", "Z") or ch in "%":
            hasil.append(ch)
        else:
            hasil.append(" ")

    text = "".join(hasil)

    # Rapikan spasi
    text = re.sub(r"\s+", " ", text).strip()

    return text

df_cleaning["clean_text_v2"] = (
    df_cleaning["original_text"].apply(clean_text_v2)
)

display(
    df_cleaning[
        ["original_text", "clean_text", "clean_text_v2"]
    ].sample(min(10, len(df_cleaning)), random_state=42)
)
```

```python
import re
import unicodedata
import pandas as pd
import html

def clean_text_v3(text):
    if pd.isna(text):
        return ""

    text = html.unescape(str(text))
    text = unicodedata.normalize("NFKC", text).lower()

    # Hapus URL dan email
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)
    text = re.sub(
        r"\b[\w.+-]+@[\w.-]+\.[a-zA-Z]{2,}\b",
        " ",
        text
    )

    # Hapus mention, pertahankan kata dari hashtag
    text = re.sub(r"(?<!\w)@\w+", " ", text)
    text = re.sub(r"#(\w+)", r"\1", text)

    # Hapus artefak encoding yang sering muncul pada emoji rusak.
    # Jangan menghapus huruf atau angka secara umum.
    text = re.sub(r"[âðï�]+", " ", text)

    # Pertahankan huruf, angka, spasi, dan tanda baca
    # yang berguna untuk teks promosi.
    hasil = []
    for ch in text:
        kategori = unicodedata.category(ch)
        if kategori[0] in ("L", "N", "P", "Z") or ch in "%":
            hasil.append(ch)
        else:
            hasil.append(" ")

    text = "".join(hasil)
    text = re.sub(r"\s+", " ", text).strip()

    return text

df_cleaning["clean_text_v3"] = (
    df_cleaning["original_text"].apply(clean_text_v3)
)

display(
    df_cleaning[
        ["original_text", "clean_text_v2", "clean_text_v3"]
    ].sample(min(10, len(df_cleaning)), random_state=42)
)
```

```python
import re
import pandas as pd

kamus_normalisasi = {
    "udah": "sudah",
    "aja": "saja",
    "gak": "tidak",
    "ga": "tidak",
    "nggak": "tidak",
    "enggak": "tidak",
    "bgt": "banget",
    "ntar": "nanti",
    "yg": "yang",
    "byk": "banyak",
    "dgn": "dengan",
    "gratong": "gratis ongkir",
    "sampe": "sampai",
    "ampe": "sampai",
    "kalo": "kalau",
    "gitu": "begitu",
    "gini": "begini",
    "onlen": "online",
    "aplgi": "apalagi",
    "dpt": "dapat",
    "jg": "juga",
    "cm": "cuma",
    "tpi": "tapi",
    "gbsa": "tidak bisa",
}

def normalize_text(text):
    if pd.isna(text):
        return ""

    text = str(text)

    # Hilangkan dekorasi berupa deretan tanda hubung
    text = re.sub(r"[-]{3,}", " ", text)

    for kata_asli, kata_baru in kamus_normalisasi.items():
        pola = r"(?<!\w)" + re.escape(kata_asli) + r"(?!\w)"
        text = re.sub(pola, kata_baru, text)

    return re.sub(r"\s+", " ", text).strip()

df_cleaning["normalized_text"] = (
    df_cleaning["clean_text_v3"].apply(normalize_text)
)

display(
    df_cleaning[
        ["original_text", "clean_text_v3", "normalized_text"]
    ].sample(min(10, len(df_cleaning)), random_state=7)
)
```

```python

import re
import pandas as pd

def label_relevansi_otomatis(teks):
    teks = str(teks).lower()

    # Deteksi penyebutan tanggal kembar / 10.10
    langsung = bool(re.search(
        r"10\s*[./-]\s*10|10\.10|tanggal\s+kembar|"
        r"\b10\s*10\b|double\s*date",
        teks
    ))

    # Indikasi percakapan tentang promosi belanja
    promo = bool(re.search(
        r"promo|promosi|diskon|discount|voucher|kupon|"
        r"flash\s*sale|flashsale|gratis\s*ongkir|freeong|"
        r"belanja|checkout|marketplace|shopee|tokopedia|"
        r"lazada|blibli|paylater|potongan\s+harga",
        teks
    ))

    # Kemungkinan penyebutan tanggal kembar yang tidak berkaitan
    konteks_tidak_relevan = bool(re.search(
        r"anaknya kembar|menikah di tanggal kembar|"
        r"ulang tahun|anniversary",
        teks
    ))

    if konteks_tidak_relevan and not promo:
        return "tidak_relevan"
    elif langsung and promo:
        return "relevan_langsung"
    elif langsung:
        return "perlu_konteks"
    elif promo:
        return "relevan_kontekstual"
    else:
        return "perlu_konteks"


# Isi hanya label yang masih kosong
mask_kosong = (
    df_cleaning["relevansi_manual"].isna()
    | df_cleaning["relevansi_manual"].astype(str).str.strip().eq("")
)

df_cleaning.loc[mask_kosong, "relevansi_manual"] = (
    df_cleaning.loc[mask_kosong, "original_text"]
    .apply(label_relevansi_otomatis)
)

print("Total tweet:", len(df_cleaning))
print("\nJumlah label:")
display(df_cleaning["relevansi_manual"].value_counts().to_frame("jumlah"))

print("\nContoh hasil label:")
display(
    df_cleaning[
        ["original_text", "relevansi_manual"]
    ].sample(min(15, len(df_cleaning)), random_state=7)
)
```

Stopword removal

```python
!pip -q install Sastrawi

import re
import nltk
from nltk.corpus import stopwords

nltk.download("stopwords", quiet=True)

stopword_id = set(stopwords.words("indonesian"))

# Jangan hapus kata yang dapat membalikkan makna sentimen
kata_penting = {
    "tidak", "bukan", "belum", "jangan", "kurang", "tanpa"
}
stopword_id = stopword_id - kata_penting

def hapus_stopword(teks):
    kata = str(teks).split()
    hasil = [k for k in kata if k not in stopword_id]
    return " ".join(hasil)

df_relevan["text_no_stopword"] = (
    df_relevan["normalized_text"].apply(hapus_stopword)
)

display(
    df_relevan[
        ["normalized_text", "text_no_stopword"]
    ].head(10)
)
```

```python
print("Jumlah data relevan:", len(df_relevan))
print(
    "Teks kosong setelah stopword removal:",
    df_relevan["text_no_stopword"].str.strip().eq("").sum()
)

display(
    df_relevan["text_no_stopword"]
    .head(20)
    .to_frame("teks_setelah_stopword")
)
```

Melakukan stemming dengan Sastrawi

```python
!pip -q install Sastrawi

from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

factory = StemmerFactory()
stemmer = factory.create_stemmer()

def lakukan_stemming(teks):
    if not isinstance(teks, str):
        return ""
    return stemmer.stem(teks)

df_relevan = df_relevan.copy()

df_relevan["stemmed_text"] = (
    df_relevan["text_no_stopword"].apply(lakukan_stemming)
)

display(
    df_relevan[
        ["text_no_stopword", "stemmed_text"]
    ].head(20)
)
```

```python
print("Jumlah data:", len(df_relevan))
print(
    "Teks kosong setelah stemming:",
    df_relevan["stemmed_text"].str.strip().eq("").sum()
)

display(
    df_relevan[
        ["original_text", "normalized_text", "stemmed_text"]
    ].head(10)
)
```
